### 🟢 Basic ML & Statistics

**Q1: What is the difference between supervised and unsupervised learning?**  
**A:** Supervised learning uses labeled data (input + output), e.g., classification and regression. Unsupervised learning deals with unlabeled data, e.g., clustering and dimensionality reduction.

**Q2: Explain bias-variance tradeoff.**  
**A:** Bias is error due to overly simplistic assumptions (underfitting). Variance is error due to sensitivity to small fluctuations in training data (overfitting). The tradeoff balances underfitting and overfitting.

**Q3: What’s the difference between Type I and Type II errors?**  
**A:** Type I (False Positive): incorrectly rejecting a true null hypothesis. Type II (False Negative): failing to reject a false null hypothesis.

**Q4: Why is A/B testing important in data science?**  
**A:** It allows for comparing two versions of a model or product feature to determine which performs better using statistical significance.

**Q5: What is regularization and why is it used?**  
**A:** Regularization penalizes large coefficients to prevent overfitting. Common types: L1 (Lasso) and L2 (Ridge).

### 🟡 Intermediate ML Concepts

**Q6: What happens if we use a very high learning rate in gradient descent?**  
**A:** The model may overshoot the optimal solution and fail to converge, bouncing around minima or diverging entirely.

**Q7: What is the curse of dimensionality?**  
**A:** As dimensions increase, data becomes sparse, distances become less meaningful, and models suffer from overfitting and increased computation.

**Q8: How do you handle missing data in a dataset?**  
**A:** Options include dropping rows/columns, imputing mean/median/mode, using algorithms that support missing values (e.g., XGBoost), or modeling missingness itself.

**Q9: When would you prefer decision trees over random forests?**  
**A:** When interpretability is crucial, and the dataset is small enough to avoid overfitting.

**Q10: What is cross-validation and why is it important?**  
**A:** Cross-validation splits data into multiple folds to evaluate model performance robustly and reduce variance in performance estimates.

### 🔵 Advanced ML & Tricky Questions

**Q11: If you have class imbalance in your data, how can you improve model performance?**  
**A:** Use metrics like F1-score, ROC-AUC; apply resampling techniques (oversampling, undersampling); use class weights; try anomaly detection methods.

**Q12: What is the difference between bagging and boosting?**  
**A:** Bagging (e.g., Random Forest) builds models in parallel and averages results. Boosting (e.g., AdaBoost, XGBoost) builds models sequentially, focusing on correcting previous errors.

**Q13: Can you have too many trees in a random forest?**  
**A:** Adding more trees doesn't cause overfitting but increases computation time. Performance stabilizes after a certain number.

**Q14: What is the purpose of PCA?**  
**A:** Principal Component Analysis reduces dimensionality by projecting data onto orthogonal components that capture maximum variance.

**Q15: Why shouldn’t you scale target variables in regression problems?**  
**A:** Scaling target variables affects interpretation and evaluation metrics (like RMSE). Only features should be scaled unless using specific loss functions.

### ⚪ Feature Engineering & Selection

**Q16: What is one-hot encoding and when should you use it?**  
**A:** One-hot encoding converts categorical variables into binary vectors. Used when categories are nominal (no inherent order).

**Q17: What is leakage in data science and how can you detect it?**  
**A:** Leakage occurs when information from outside the training set is used during model training, leading to overly optimistic performance. Check for features derived from future data or validation/test sets.

**Q18: How do you encode high-cardinality categorical variables?**  
**A:** Use target encoding, leave-one-out encoding, embedding layers, or hashing trick.

**Q19: What is the difference between normalization and standardization?**  
**A:** Normalization scales data to [0,1], while standardization centers data around zero with unit variance. Standardization is better for Gaussian-like distributions.

**Q20: Should you do feature selection before or after splitting the data?**  
**A:** After. Doing it before causes data leakage because feature importance or selection criteria might depend on the entire dataset.

### 🟣 Model Evaluation & Interpretation

**Q21: What is the difference between precision and recall?**  
**A:** Precision = TP / (TP + FP) → how many selected items are relevant. Recall = TP / (TP + FN) → how many relevant items were selected.

**Q22: In which scenario is accuracy not a good metric?**  
**A:** When dealing with imbalanced datasets. For example, 99% negative samples — a model predicting all negatives will have 99% accuracy but is useless.

**Q23: What is the AUC-ROC curve and how does it work?**  
**A:** AUC measures the area under the ROC curve, showing the model’s ability to distinguish classes at different thresholds. Higher AUC = better performance.

**Q24: What is log loss and how is it used?**  
**A:** Log loss penalizes confident wrong predictions more severely. Used as a loss function in logistic regression and probabilistic classifiers.

**Q25: How can you explain the predictions of a black-box model like XGBoost or neural networks?**  
**A:** Use SHAP (SHapley Additive exPlanations), LIME (Local Interpretable Model-agnostic Explanations), or permutation importance.

### 🔴 Deep Learning & NLP (Advanced)

**Q26: What is the vanishing gradient problem and how can it be mitigated?**  
**A:** Gradients become very small during backpropagation, slowing or stopping learning. Mitigated using ReLU activation, batch normalization, or LSTM in RNNs.

**Q27: What is dropout in neural networks?**  
**A:** Dropout randomly disables neurons during training to prevent overfitting and improve generalization.

**Q28: What is transfer learning and how is it used in NLP?**  
**A:** Transfer learning involves using pre-trained models (e.g., BERT, GPT) and fine-tuning them on domain-specific tasks to leverage learned representations.

**Q29: What is the difference between stemming and lemmatization?**  
**A:** Stemming cuts words to their root form (may not be a real word). Lemmatization reduces words to dictionary/base forms using context.

**Q30: What is attention in transformers and why is it important?**  
**A:** Attention mechanisms allow models to focus on relevant parts of input sequences, enabling handling of long-range dependencies efficiently.

### ⚫ Real-World Scenarios & Case Studies

**Q31: You're given a dataset with 1 million rows and 1000 features. How would you approach building a model?**  
**A:** Start with exploratory data analysis (EDA), check for missing values, perform feature selection (e.g., correlation matrix, variance threshold), apply dimensionality reduction, choose appropriate models (e.g., tree-based or linear), and optimize with cross-validation.

**Q32: Your model performs well on training data but poorly on test data. What could be the issue?**  
**A:** Overfitting. Consider regularization, simplifying the model, adding more data, or using cross-validation.

**Q33: You deploy a model in production, and its performance drops significantly after a month. Why?**  
**A:** Data drift (distribution shift), concept drift (target changes over time), or issues in preprocessing/data pipeline.

**Q34: How do you monitor a deployed machine learning model?**  
**A:** Track prediction drift, model performance (if labels come in later), system health, latency, and business KPIs.

**Q35: You have a model with 99% accuracy. Is it necessarily good?**  
**A:** Not necessarily. Could be an imbalanced dataset (e.g., 99% of one class). Look at other metrics like F1, precision, recall, confusion matrix.

### 🟠 Probability & Statistics

**Q36: What is the Central Limit Theorem (CLT)?**  
**A:** CLT states that the sampling distribution of the sample mean approaches a normal distribution as the sample size increases, regardless of population distribution.

**Q37: What is p-value?**  
**A:** The probability of observing the data (or more extreme) assuming the null hypothesis is true. Low p-values suggest evidence against the null.

**Q38: What is the difference between covariance and correlation?**  
**A:** Covariance shows direction of relationship, but not strength. Correlation is standardized covariance (range -1 to 1), indicating both direction and strength.

**Q39: If two variables are independent, what is their covariance?**  
**A:** Zero. However, zero covariance does not always imply independence.

**Q40: What is Simpson’s Paradox?**  
**A:** A phenomenon where a trend appears in several groups but disappears or reverses when groups are combined.

**Q42: How do you handle categorical variables in scikit-learn pipelines?**  
**A:** Use `ColumnTransformer` with `OneHotEncoder`, `SimpleImputer`, etc., inside a `Pipeline`.

**Q43: What is the difference between `fit()` and `fit_transform()` in scikit-learn?**  
**A:** `fit()` learns parameters from data. `fit_transform()` fits the data and returns transformed data (used during training). `transform()` applies transformation without fitting (used during inference).

**Q44: How can you speed up hyperparameter tuning?**  
**A:** Use `RandomizedSearchCV` instead of `GridSearchCV`, Bayesian optimization, early stopping, or reduce search space.

**Q45: What is stratified k-fold cross-validation and when should you use it?**  
**A:** Maintains class distribution in each fold. Use for classification problems with imbalanced classes.
