### When to Use

When you want to measure the overall quality of clustering.

For algorithms like K-Means, Hierarchical Clustering, or DBSCAN.

```python
from sklearn.metrics import silhouette_score

# Compute Silhouette Score
score = silhouette_score(X, labels)
print(f'Silhouette Score: {score}')
```

### Interpretation

A high score (> 0.5) indicates good clustering.

A low score (< 0) suggests poor clustering or overlapping clusters.

```python
from sklearn.metrics import davies_bouldin_score

# Compute Davies-Bouldin Index
db_index = davies_bouldin_score(X, labels)
print(f'Davies-Bouldin Index: {db_index}')
```

###Interpretation

Lower values are better (closer to 0).

Higher values indicate overlapping or poorly separated clusters.

image.png

```python
from sklearn.metrics import calinski_harabasz_score

# Compute Calinski-Harabasz Index
ch_score = calinski_harabasz_score(X, labels)
print(f'Calinski-Harabasz Index: {ch_score}')
```

### Interpretation
Higher values are better.

Works well for spherical clusters but may struggle with non-spherical shapes.

image.png

```python
from sklearn.metrics import adjusted_rand_score

# Ground truth labels (if available)
true_labels = [0, 1, 1, 0, 2, 2]

# Compute ARI
ari_score = adjusted_rand_score(true_labels, labels)
print(f'Adjusted Rand Index: {ari_score}')
```

### Interpretation
Ranges from -1 to 1 :

1 : Perfect match.

0 : Random labeling.

-1 : Anti-correlation.

```python
from sklearn.cluster import KMeans

# Compute inertia for different values of k
inertia = []
for k in range(1, 11):
    kmeans = KMeans(n_clusters=k, random_state=0)
    kmeans.fit(X)
    inertia.append(kmeans.inertia_)

# Plot Elbow curve
import matplotlib.pyplot as plt
plt.plot(range(1, 11), inertia, marker='o')
plt.xlabel('Number of Clusters (k)')
plt.ylabel('Inertia')
plt.title('Elbow Method')
plt.show()
```

###Interpretation

Look for the "elbow point" where the rate of decrease in inertia slows down.
