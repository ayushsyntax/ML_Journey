L1 vs. L2 Regularization (Lasso vs. Ridge) – Everything You Need to Know
L1 (Lasso) and L2 (Ridge) regularization are techniques used to prevent overfitting in machine learning models by penalizing large coefficients. While they seem similar, they behave differently in feature selection, optimization, and performance.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import Lasso, Ridge
from sklearn.datasets import make_regression

# Generate synthetic dataset
X, y = make_regression(n_samples=100, n_features=10, noise=5, random_state=42)

# Apply Lasso (L1)
lasso = Lasso(alpha=0.1)
lasso.fit(X, y)
print("Lasso coefficients:", lasso.coef_)  # Some will be zero

# Apply Ridge (L2)
ridge = Ridge(alpha=0.1)
ridge.fit(X, y)
print("Ridge coefficients:", ridge.coef_)  # All will be small but nonzero
```

7️⃣ Final Takeaways
✅ L1 (Lasso) → Feature selection, sparse models, some weights = 0.
✅ L2 (Ridge) → Keeps all features, prevents overfitting, better for multicollinearity.
✅ Elastic Net = L1 + L2 for best of both worlds.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import Lasso, Ridge
from sklearn.datasets import make_regression

# Generate synthetic dataset with 100 samples and 10 features
X, y = make_regression(n_samples=100, n_features=10, noise=5, random_state=42)

# Apply Lasso (L1)
lasso = Lasso(alpha=0.1)
lasso.fit(X, y)
lasso_coefs = lasso.coef_

# Apply Ridge (L2)
ridge = Ridge(alpha=0.1)
ridge.fit(X, y)
ridge_coefs = ridge.coef_

# Plot comparison
plt.figure(figsize=(10, 5))
plt.plot(lasso_coefs, 'ro-', label="Lasso (L1)", markersize=8)  # Red dots for L1
plt.plot(ridge_coefs, 'bo-', label="Ridge (L2)", markersize=8)  # Blue dots for L2

# Highlight zero coefficients in Lasso
plt.axhline(y=0, color='gray', linestyle='--', linewidth=1)
plt.xlabel("Feature Index")
plt.ylabel("Coefficient Value")
plt.title("L1 (Lasso) vs. L2 (Ridge) Regularization Effect on Coefficients")
plt.legend()
plt.grid(True)
plt.show()
```
