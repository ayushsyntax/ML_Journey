```python
from sklearn.metrics import accuracy_score

y_true = [[1, 0, 1], [0, 1, 0]]
y_pred = [[1, 1, 0], [0, 1, 1]]

exact_match = accuracy_score(y_true, y_pred)
print(f"Exact Match: {exact_match}")  # Output: 0.0 (0/2 correct)
```

When to Use:

When all labels must be correct (strict evaluation).

Example: Medical diagnosis where every predicted condition must match reality


image.png

```python
from sklearn.metrics import hamming_loss

h_loss = hamming_loss(y_true, y_pred)
print(f"Hamming Loss: {h_loss}")  # Output: 0.5 (3 errors / 6 labels)
```

When to Use:

When partial correctness matters.

Example: Tagging systems (e.g., predicting multiple hashtags for a post).


image.png

```python
from sklearn.metrics import f1_score

f1_macro = f1_score(y_true, y_pred, average='macro')  # 0.555
f1_micro = f1_score(y_true, y_pred, average='micro')  # 0.5

print(f"Macro-F1: {f1_macro}, Micro-F1: {f1_micro}")
```

When to Use:

Macro-F1: When all labels are equally important (e.g., rare diseases).

Micro-F1: When frequent labels dominate (e.g., spam detection).

```python
from sklearn.metrics import jaccard_score

jaccard = jaccard_score(y_true, y_pred, average='samples')  # 0.25
print(f"Jaccard Similarity: {jaccard}")
```

When to Use:

When partial matches are acceptable.

Example: Image segmentation (IoU is standard in object detection).

```python
from sklearn.metrics import roc_auc_score
import numpy as np

y_true = np.array([[1, 0, 1], [0, 1, 0]])
y_proba = np.array([[0.9, 0.4, 0.6], [0.1, 0.8, 0.3]])

roc_auc_macro = roc_auc_score(y_true, y_proba, average='macro')  # 0.833
roc_auc_micro = roc_auc_score(y_true.ravel(), y_proba.ravel())  # 0.8
```

When to Use:

When probabilistic predictions are available.

Example: Fraud detection (ranking suspicious transactions)
