**Fairness in Machine Learning (ML)** is a critical area of research and practice that focuses on ensuring that machine learning models do not reinforce or amplify existing societal biases. As ML systems are increasingly used to make decisions in high-stakes domains like hiring, lending, healthcare, criminal justice, and education, ensuring fairness becomes essential for ethical, legal, and practical reasons.

---

## 📌 **Core Idea of Fairness in ML**

The core idea behind **fairness in machine learning** is to ensure that the model's predictions or decisions do **not disproportionately disadvantage individuals or groups based on sensitive attributes**, such as:

- Race
- Gender
- Age
- Religion
- Disability
- Nationality
- Sexual orientation

A fair model should treat all individuals equally or similarly under similar conditions, regardless of these protected characteristics.

---

## 🔍 **Why Does Bias Occur in ML Models?**

Bias can enter an ML system at multiple stages:

### 1. **Data Collection**
- Historical bias: Data reflects past discrimination.
- Representation bias: Underrepresentation of certain groups.
- Measurement bias: Features may be measured differently across groups.

### 2. **Feature Selection**
- Biased proxies: Using features that act as proxies for sensitive attributes (e.g., zip code as proxy for race).

### 3. **Model Training**
- Optimization for accuracy over fairness.
- Reinforcement of patterns in biased data.

### 4. **Evaluation**
- Lack of fairness-aware evaluation metrics.

### 5. **Deployment**
- Contextual mismatch: Model used in settings where it was not tested for fairness.

---

## 🧩 **Types of Fairness in ML**

There are several mathematical formalizations of fairness, each with its own assumptions and trade-offs. The main types include:

### 1. **Group Fairness (Statistical Fairness)**
Ensures that outcomes are distributed similarly across different groups.

#### Examples:
- **Demographic Parity (Statistical Parity):**  
  The probability of a positive outcome is the same across all groups.  
  $ P(\hat{Y} = 1 | A = a) = P(\hat{Y} = 1 | A = b) $

- **Equal Opportunity:**  
  True positive rates are equal across groups.  
  $ P(\hat{Y} = 1 | Y = 1, A = a) = P(\hat{Y} = 1 | Y = 1, A = b) $

- **Equalized Odds:**  
  Both true positive and false positive rates are equal across groups.  
  $ P(\hat{Y} | Y, A = a) = P(\hat{Y} | Y, A = b) $

- **Predictive Parity:**  
  Positive predictive values are equal across groups.  
  $ P(Y = 1 | \hat{Y} = 1, A = a) = P(Y = 1 | \hat{Y} = 1, A = b) $

> ⚠️ **Note:** These criteria often conflict; you cannot satisfy them all simultaneously unless under specific conditions (see "Impossibility Theorem" by Kleinberg et al.).

---

### 2. **Individual Fairness**
Treats similar individuals similarly.

- Requires a **task-specific similarity metric** between individuals.
- Formalization: If two individuals are similar, their predicted outcomes should also be similar.

$$
d(x_1, x_2) < \epsilon \Rightarrow |\hat{Y}(x_1) - \hat{Y}(x_2)| < \delta
$$

Where $ d $ is a distance metric capturing similarity.

---

### 3. **Causal Fairness**
Considers causal relationships between variables to identify and mitigate bias.

- Involves counterfactual reasoning: Would the decision change if the person belonged to a different group?
- Tools like **causal graphs** and **interventional analysis** are used.

Examples:
- **Counterfactual Fairness**
- **No Direct/Indirect Discrimination**

## 🛠️ **When Does Fairness Come Into Play?**

Fairness should be considered throughout the ML lifecycle:

| Stage | Considerations |
|-------|----------------|
| **Problem Definition** | Define what fairness means in context. Identify protected attributes. |
| **Data Collection** | Ensure representative and unbiased data. |
| **Preprocessing** | Remove bias from training data (e.g., reweighting, resampling). |
| **Model Training** | Use fairness-aware algorithms or constraints. |
| **Evaluation** | Measure both performance and fairness metrics. |
| **Deployment** | Monitor model behavior in production for unintended consequences. |
| **Feedback Loop** | Update model regularly and audit for fairness drift. |

---

## 🧰 **How Do We Deal With Unfairness in ML?**

There are three broad strategies to address fairness:

### 1. **Preprocessing Techniques**
Modify the training data before model training to remove bias.

- Reweighting instances
- Resampling
- Transforming features
- Adversarial debiasing

### 2. **In-processing Techniques**
Modify the learning algorithm to incorporate fairness constraints during training.

- Regularization for fairness
- Constrained optimization
- Adversarial training
- Fair representation learning

### 3. **Post-processing Techniques**
Adjust the model’s output after training to achieve fairness.

- Calibration
- Threshold adjustment
- Reject option-based classification

## 📉 Real-Life Scenario: **Amazon’s AI Recruitment Tool Bias (2014–2017)**

### 🔍 Background:

In 2014, **Amazon** started building an AI-powered recruitment tool to help automate the screening of job applicants. The goal was to improve hiring efficiency by using machine learning to rate resumes on a scale from 1 to 5 stars.

### ⚠️ Problem Identified:

By 2015, Amazon realized that the model was **systematically downgrading resumes that included words like “women” or all-female colleges**. For example:
- Resumes containing “women’s chess club captain” were rated lower.
- Graduates from all-women universities were penalized.

The system had learned from **historical hiring data**, which reflected past biases — mostly male hires over a 10-year period.

### ❌ Why It Happened:

1. **Biased Training Data:**  
   - Most training data came from resumes submitted over 10 years, during which tech hiring was male-dominated.
   - The model associated maleness with higher scores.

2. **Proxy Discrimination:**  
   - Words like “women’s” acted as proxies for gender, even though gender wasn’t explicitly used as a feature.

3. **Lack of Fairness Evaluation:**  
   - No fairness-aware checks were built into the model evaluation process.

### 🧱 What They Tried:

- Amazon tried to retrain the model without gender-related keywords.
- However, the model still found **other proxies** for gender (e.g., all-women clubs, certain universities).

Eventually, Amazon **scrapped the project in 2017**, realizing it couldn't fully eliminate bias.

---

## 🛠️ How This Could Have Been Handled (Fairness in ML Approach):

Had Amazon applied **fairness-aware ML techniques**, they might have mitigated this issue.

Here's how:

### ✅ Step-by-Step Fairness Fix (Hypothetical Solution Using `Fairlearn`):

```python
from fairlearn.postprocessing import ThresholdOptimizer
from sklearn.ensemble import RandomForestClassifier
from fairlearn.metrics import MetricFrame
from sklearn.metrics import accuracy_score

# Assume X_train, y_train, A_train = resume features, labels, gender

# Train initial model
model = RandomForestClassifier().fit(X_train, y_train)

# Evaluate fairness
y_pred = model.predict(X_test)

# Apply post-processing for fairness
fair_model = ThresholdOptimizer(estimator=model, constraints="equalized_odds", prefit=True)
fair_model.fit(X_train, y_train, sensitive_features=A_train)

# Predict and evaluate again
y_fair = fair_model.predict(X_test, sensitive_features=A_test)
mf_fair = MetricFrame(accuracy_score, y_test, y_fair, sensitive_features=A_test)
print("After Fairness Fix - Accuracy by group:", mf_fair.by_group)
```

This would have helped **equalize performance across genders**, reducing bias while maintaining utility.
