```python
# Importing libraries
from sklearn.neighbors import KNeighborsClassifier
import numpy as np

# Data: [Weight (g), Size (cm)]
X = np.array([
    [150, 7], [170, 8],  # Apples (Class 0)
    [130, 5], [120, 4]   # Oranges (Class 1)
])

# Labels (0 = Apple, 1 = Orange)
y = np.array([0, 0, 1, 1])

# New fruit to classify (160g, 6cm)
new_fruit = np.array([[160, 6]])

# KNN Classifier with K=3
knn_classifier = KNeighborsClassifier(n_neighbors=3)
knn_classifier.fit(X, y)

# Predicting the class
prediction = knn_classifier.predict(new_fruit)

# Output the prediction
print("Predicted Class (0 = Apple, 1 = Orange):", prediction[0])
```

```python
# Importing libraries
from sklearn.neighbors import KNeighborsRegressor
import numpy as np

# Data: [Size (sq ft)]
X = np.array([[1200], [1500], [1700], [2000]])

# House prices (₹ Lakhs)
y = np.array([50, 60, 80, 100])

# New house (1600 sq ft)
new_house = np.array([[1600]])

# KNN Regressor with K=3
knn_regressor = KNeighborsRegressor(n_neighbors=3)
knn_regressor.fit(X, y)

# Predicting the price
predicted_price = knn_regressor.predict(new_house)

# Output the prediction
print("Predicted House Price (₹ Lakhs):", predicted_price[0])
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import cross_val_score
import matplotlib.pyplot as plt
import numpy as np

# Example dataset
from sklearn.datasets import load_iris
X, y = load_iris(return_X_y=True)

# Test different k values
k_values = range(1, 31)  # Test k from 1 to 30
cv_scores = []

for k in k_values:
    knn = KNeighborsClassifier(n_neighbors=k)
    scores = cross_val_score(knn, X, y, cv=5, scoring='accuracy')  # 5-fold cross-validation
    cv_scores.append(scores.mean())

# Plot accuracy vs k
plt.plot(k_values, cv_scores, marker='o')
plt.xlabel('k (Number of Neighbors)')
plt.ylabel('Accuracy')
plt.title('Choosing the Best k')
plt.show()

# Find the best k
best_k = k_values[np.argmax(cv_scores)]
print(f"Best k: {best_k}")
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import cross_val_score
import matplotlib.pyplot as plt

# Example dataset
from sklearn.datasets import load_iris
X, y = load_iris(return_X_y=True)

# Test k values from 1 to 20
k_values = range(1, 21)
accuracies = []

for k in k_values:
    knn = KNeighborsClassifier(n_neighbors=k)
    scores = cross_val_score(knn, X, y, cv=5)  # 5-fold cross-validation
    accuracies.append(scores.mean())

# Find the best k
best_k = k_values[accuracies.index(max(accuracies))]
print(f"Best k: {best_k}")
```
