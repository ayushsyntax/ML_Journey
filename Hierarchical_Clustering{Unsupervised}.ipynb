```python
# Step 1: Import Libraries
import numpy as np
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import dendrogram, linkage
from sklearn.datasets import make_blobs

# Step 2: Generate Sample Data
X, _ = make_blobs(n_samples=100, centers=3, random_state=42)

# Step 3: Perform Hierarchical Clustering
Z = linkage(X, method='ward')  # 'ward' minimizes variance between clusters

# Step 4: Visualize the Dendrogram
plt.figure(figsize=(10, 6))
plt.title("Hierarchical Clustering Dendrogram")
plt.xlabel("Data Points")
plt.ylabel("Distance")
dendrogram(Z, truncate_mode='lastp', p=10, leaf_rotation=45, leaf_font_size=10)
plt.show()

# Step 5: Interpret the Results
print("Dendrogram shows how clusters are merged at each step.")
```
