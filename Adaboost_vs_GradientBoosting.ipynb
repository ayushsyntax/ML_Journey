

---

### **1. Leaf Nodes**
- **AdaBoost**:
  - Typically uses **shallow decision trees** (e.g., decision stumps with only one split, i.e., two leaf nodes).
  - Focuses on simplicity to reduce overfitting since the model relies heavily on weighting and sequential corrections.

- **Gradient Boosting**:
  - Can use **deeper trees** with more leaf nodes (e.g., `max_depth=3` or higher).
  - The depth of the tree is a hyperparameter that controls the complexity of each weak learner.

---

### **2. Learning Rate**
- **AdaBoost**:
  - Does not explicitly use a **learning rate**. Instead, it adjusts the contribution of each weak learner by assigning **weights** to data points based on errors.
  - Misclassified points are given higher weights, forcing subsequent models to focus on them.

- **Gradient Boosting**:
  - Explicitly uses a **learning rate** (`lr`) to scale the contribution of each weak learner.
  - A smaller learning rate slows down learning but improves generalization by allowing for finer adjustments.

---

### **3. Weightage**
- **AdaBoost**:
  - Assigns **weights to data points**. After each iteration:
    - Correctly classified points have their weights reduced.
    - Misclassified points have their weights increased.
  - The final prediction is a weighted sum of all weak learners, where the weights are proportional to the accuracy of each learner.

- **Gradient Boosting**:
  - Assigns **weights to models**, controlled by the learning rate.
  - Each new model focuses on correcting the **residuals** (errors) of the previous models, rather than adjusting data point weights.
  - All models contribute equally to the final prediction (scaled by the learning rate).

---

### **4. Training Process**
- **AdaBoost**:
  - Trains each weak learner sequentially, focusing on the **misclassified points** from the previous learner.
  - Errors are minimized by adjusting the **weights of data points**.

- **Gradient Boosting**:
  - Trains each weak learner sequentially, focusing on the **residuals** (errors) of the previous learner.
  - Errors are minimized by fitting new models to the residuals.

---

### **5. Loss Function**
- **AdaBoost**:
  - Implicitly minimizes an **exponential loss function** by adjusting weights on misclassified points.
  - Works well for classification tasks.

- **Gradient Boosting**:
  - Explicitly minimizes a **differentiable loss function** (e.g., Mean Squared Error for regression, Log Loss for classification).
  - More flexible and can handle various types of problems.

---

### **6. Robustness**
- **AdaBoost**:
  - Sensitive to noisy data and outliers because it heavily weights misclassified points, which can lead to overfitting.

- **Gradient Boosting**:
  - More robust to noise due to the use of a learning rate and deeper trees, which allow for smoother corrections.

---

### **Summary Table**

| Feature               | AdaBoost                                   | Gradient Boosting                          |
|-----------------------|--------------------------------------------|-------------------------------------------|
| **Leaf Nodes**        | Shallow trees (e.g., decision stumps)      | Deeper trees (configurable `max_depth`)   |
| **Learning Rate**     | No explicit learning rate                  | Explicit learning rate (`lr`)             |
| **Weightage**         | Weights data points                       | Weights models (via residuals)            |
| **Training Focus**    | Misclassified points                      | Residuals (errors)                        |
| **Loss Function**     | Exponential loss                          | General differentiable loss function      |
| **Robustness**        | Sensitive to noise                        | More robust to noise                      |

---

### **Key Takeaway**
- **AdaBoost**: Adjusts **data point weights** to focus on hard-to-classify points, using shallow trees and no explicit learning rate.
- **Gradient Boosting**: Fits new models to **residuals**, using deeper trees and an explicit learning rate to control model contributions.

$$
\boxed{\text{AdaBoost focuses on weighted data points; Gradient Boosting focuses on residuals with a learning rate.}}
$$
