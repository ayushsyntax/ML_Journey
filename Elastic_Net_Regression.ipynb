```python
from sklearn.datasets import load_diabetes
from sklearn.linear_model import LinearRegression,Ridge,Lasso,ElasticNet
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
```

```python
X,y = load_diabetes(return_X_y = True)
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size = 0.3,random_state = 42)
```

```python
#linear Regression
reg = LinearRegression()
reg.fit(X_train,y_train)
y_pred = reg.predict(X_test)
r2_score(y_test,y_pred)
```

```python
#Ridge regression(L2)
reg = Ridge(alpha = 0.2)
reg.fit(X_train,y_train)
y_pred = reg.predict(X_test)
r2_score(y_test,y_pred)
```

```python
#Lasso(L1)
reg = Lasso(alpha = 0.2)
reg.fit(X_train,y_train)
y_pred = reg.predict(X_test)
r2_score(y_test,y_pred)
```

```python
#ElasticNet
reg = ElasticNet(alpha = 0.2)
reg.fit(X_train,y_train)
y_pred = reg.predict(X_test)
r2_score(y_test,y_pred)
```

**Elastic Net** is a regularization technique that combines Lasso (L1 regularization) and Ridge (L2 regularization), aiming to get the best of both worlds.

```python
from sklearn.linear_model import ElasticNet
from sklearn.model_selection import GridSearchCV

# Create Elastic Net model
elastic_net = ElasticNet(alpha=0.5, l1_ratio=0.5)  # Adjust alpha and l1_ratio

# Fit model to training data
elastic_net.fit(X_train, y_train)

# Get coefficients
print(elastic_net.coef_)
```

```python
param_grid = {'alpha': [0.1, 1, 10], 'l1_ratio': [0.1, 0.5, 0.9]}
grid_search = GridSearchCV(ElasticNet(), param_grid, cv=5)
grid_search.fit(X_train, y_train)

# Best parameters
print(grid_search.best_params_)
```

Conclusion
✅ Elastic Net = Lasso + Ridge
✅ Sparsity + Stability
✅ Great for high-dimensional & correlated data

🚀 Use Elastic Net when you need feature selection + robustness to multicollinearity!
